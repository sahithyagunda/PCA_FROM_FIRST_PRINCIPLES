```python
import numpy as np
```

```python
data = [[2,3],[4,5],[6,7]]
```

```python
## compute the mean of each feature
n = len(data)
d = len(data[1])
mean = [0]*d

for j in range(d):
    for i in range(n):
        mean[j] +=data[i][j] 
    mean[j] /= n
```

```python
print(mean)
```

```python
## mean centred 
mean_centred = []

for i in range(n):
    row = []
    for j in range(d):
        row.append(data[i][j] - mean[j])
    mean_centred.append(row)
print(mean_centred)
```

```python
### covariance matrix
import numpy as np
cov_matrix = np.zeros((d,d))

for i in range(d):
    for j in range(d):
        for k in range(n):
            cov_matrix[i][j] += mean_centred[k][i] * mean_centred[k][j] 
        cov_matrix[i][j] /= (n-1)
print(cov_matrix)


```

```python
## finding largest eigen value and eigen vector using power iteration

v = [1,0]

def mat_vec_mult(cov_matrix,v):
    resultant_matrix = [0]* len(v)
    for i in range(len(cov_matrix)):
        for j in range(len(cov_matrix)):
            resultant_matrix[i] += cov_matrix[i][j] * v[j]
    return resultant_matrix
print(mat_vec_mult(cov_matrix,v))
```

```python
## magnitude of vector
import math
def magnitude(resultant_matrix):
    mag = 0
    for i in range(len(resultant_matrix)):
        mag += resultant_matrix[i]* resultant_matrix[i]
    return math.sqrt(mag)
print(magnitude(mat_vec_mult(cov_matrix,v)))
```

```python
for iters in range(100):
    v_new = mat_vec_mult(cov_matrix,v)
    mag = magnitude(v_new)
    for i in range(d):
        v[i] = v_new[i]/mag
print("Eigen Vector:",v)
```

```python
## computing eigen value
lambda_value = 0
temp = mat_vec_mult(cov_matrix,v)
for i in range(d):
    lambda_value += v[i]*temp[i]
print("Eigen Value:",lambda_value)
```

```python
## deflation to find next eigen value and eigen vector
new_cov_matrix = [[0 for _ in range(d)] for _ in range(d)]
for i in range(d):
    for j in range(d):
        new_cov_matrix[i][j] = cov_matrix[i][j] - lambda_value * v[i] * v[j]
print(new_cov_matrix)
```

```python
#### projection
z = [0] * n

for i in range(n):
    for j in range(d):
        z[i] += mean_centred[i][j] * v[j]
print("Projected Data:",z)
```
